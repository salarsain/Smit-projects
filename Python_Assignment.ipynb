## Q1: Positive, Negative, or Zero

```
num = float(input("Enter a number: "))

if num > 0:
    print("The number is Positive")
elif num < 0:
    print("The number is Negative")
else:
    print("The number is Zero")
```

## Q2: Even or Odd

```
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
```

## Q3: Sum, Difference, Product, Quotient

```
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Sum:", a + b)
print("Difference:", a - b)
print("Product:", a * b)

if b != 0:
    print("Quotient:", a / b)
else:
    print("Cannot divide by zero")
```

## Q4: Grade Calculation

```
marks = int(input("Enter marks (0–100): "))

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

## Q5: Largest of Three Numbers

```
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))

if a >= b and a >= c:
    print("Largest number is:", a)
elif b >= a and b >= c:
    print("Largest number is:", b)
else:
    print("Largest number is:", c)
```

## Q6: Temperature Classification

```
temp = float(input("Enter current temperature: "))

if temp >= 30:
    print("Hot")
elif temp >= 20:
    print("Warm")
else:
    print("Cold")
```

## Q7: Divisible by 5

```
num = int(input("Enter a number: "))

if num % 5 == 0:
    print("Number is divisible by 5")
else:
    print("Not divisible by 5")
```

## Q8: Bill Discount

```
bill = float(input("Enter bill amount: "))

if bill >= 1000:
    discount = bill * 0.10
    final_amount = bill - discount
else:
    final_amount = bill

print("Final bill amount:", final_amount)
```

## Q9: Leap Year Check

```
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

## Q10: Login System

```
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin" and password == "1234":
    print("Login Successful")
else:
    print("Login Failed")
```

## Q11: Square or Rectangle

```
length = float(input("Enter length: "))
width = float(input("Enter width: "))

if length == width:
    print("It is a Square")
else:
    print("It is a Rectangle")
```
